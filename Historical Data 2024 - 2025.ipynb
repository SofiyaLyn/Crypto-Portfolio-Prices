```python
import pandas as pd
import yfinance as yf

tickers = [
    "AKT-USD","API3-USD","ARB11841-USD","ASTRO23374-USD","ATOM-USD",
    "AVAX-USD","AXL17799-USD","BTC-USD","CRV-USD","CVX-USD","DVPN-USD",
    "ETH-USD","FET-USD","FIL-USD","INJ-USD","JTO-USD","JUNO-USD",
    "JUP-USD","KMNO-USD","KUJI-USD","LINK-USD","LUNC-USD","LUNA20314-USD",
    "NEAR-USD","OSMO-USD","PYTH-USD","RUNE-USD","SCRT-USD","SHD-USD",
    "SNX-USD","SOMM-USD","STARS16842-USD","STRD-USD","SYN12147-USD",
    "TENEO-USD","TON11419-USD","USDC-USD","USDT-USD","WBTC-USD","ZETA-USD"
]
# TENEO does not exist 
START_DATE = "2024-12-30"
END_DATE = "2026-01-01"   
INTERVAL = "1h"
```

```python
raw_data = yf.download(
    tickers=tickers,
    start=START_DATE,
    end=END_DATE,
    interval=INTERVAL,
    auto_adjust=True
)

df_wide = raw_data['Close'].copy()

df_wide.index.name = "Date"
df_wide.index = pd.to_datetime(df_wide.index, utc=True).tz_localize(None)

df_wide = df_wide[
    (df_wide.index >= "2024-12-30 00:00:00") &
    (df_wide.index <= "2025-12-31 23:00:00")
]

df_wide = df_wide[~df_wide.index.duplicated(keep='first')]

data = df_wide.reset_index().melt(
    id_vars="Date", 
    var_name="Ticker", 
    value_name="Close"
)
data = data.dropna(subset=["Close"]).copy()

data = data.sort_values(by=["Ticker", "Date"])
data["Hourly_Return"] = data.groupby("Ticker")["Close"].pct_change().fillna(0)

data.to_csv("crypto_hourly_long.csv", index=False)
df_wide.to_csv("crypto_hourly_prices_wide.csv")
```

```python
print(f"Total Rows in Long Format: {len(data)}")
print(f"Total Rows in Wide Format: {len(df_wide)}")
```
